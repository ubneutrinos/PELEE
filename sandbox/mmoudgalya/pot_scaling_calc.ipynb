```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
samples = ["run1_nue", "run2_nue", "run3_nue", "run4a_nue", "run4b_nue", "run4c_nue", "run4d_nue", "run5_nue", "run1A_nue", "run1B_nue"]
nue_pots = [5.22325E+22, 6.41637E+22, 7.77356E+22, 3.86441E+22, 1.14936E+23, 6.88111E+22, 1.39733E+23, 1.59275E+23, 5.22325E+22, 5.22325E+22]
data_pots = [1.668E+20, 2.612E+20, 2.568E+20, 4.497E+19, 1.363E+20, 8.946E+19, 4.928E+19, 1.477E+20, 5.885E+19, 8.932E+19] # correct POTs
#data_pots = [1.668E+20, 2.612E+20, 2.568E+20, 4.497E+19, 1.363E+20, 8.946E+19, 5.01E+19, 1.55E+20, 5.885E+19, 8.932E+19] # incorrect POTs
n_mc_evts = [41146, 52035, 61957, 27087, 90773, 54336, 109890, 122036, 41146, 41146]
```

```python
# Old PELEE POT scaling method
exp_old = np.sum(n_mc_evts) * np.sum(data_pots) / np.sum(nue_pots)
print(exp_old)
```

```python
# New PELEE POT scaling method
exp_new = 0
for i in range(len(samples)):
    exp_new += n_mc_evts[i] * data_pots[i] / nue_pots[i]
print(exp_new)
```

```python
print(exp_new - exp_old)
```
